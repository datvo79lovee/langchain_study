# LLM CALL


```python
from langchain_ollama import ChatOllama

llm_ollama = ChatOllama(model = 'llama3.2:3b', temperature=0)

response = llm_ollama.invoke("Nói cho tôi về langchain bằng 2 câu")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model #có thể linh động đổi model
llm_ollama = init_chat_model(model = 'llama3.2:3b', model_provider='ollama')
llm_ollama.invoke('hello, bạn khỏe k').content
```

```python

import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

# Nạp biến từ file .env
load_dotenv()

# Kiểm tra đúng tên biến LangChain cần
api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError(
        "Không tìm thấy GOOGLE_API_KEY. "
        "Hãy kiểm tra file .env và thư mục hiện tại."
    )

print("Đã tìm thấy API key")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
)

response = llm.invoke("BM25 có tác dụng gì?")
print(response.content)
```
