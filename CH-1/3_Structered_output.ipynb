# Pydantic model: validate

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from pydantic import BaseModel, Field

# 1. Định nghĩa Pydantic Schema để ép kiểu kết quả đầu ra của AI
class LLMResponseSchema(BaseModel):
    topic_name: str = Field(description="Tên chủ đề được người dùng hỏi")
    summary: str = Field(description="Bản tóm tắt ngắn gọn về chủ đề đó dựa trên vai trò được giao")
    key_points: list[str] = Field(description="Danh sách các ý chính hoặc gạch đầu dòng quan trọng")
    expertise_level: str = Field(description="Đánh giá mức độ chuyên sâu của câu trả lời")

# 2. Khởi tạo LLM và gắn Structured Output theo schema
llm_ollama = ChatOllama(model='llama3.2:3b', temperature=0)
structured_llm = llm_ollama.with_structured_output(LLMResponseSchema)

# 3. Khởi tạo Prompt Template
user_prompt = ChatPromptTemplate([
    ('system', 'bạn là {role}'),
    ('user', 'hãy nói cho tôi về {topic}')
])

# 4. Nhận input
role_input = input("nhập vai trò: ")
user_input = input('nhập chủ đề muốn hỏi : ')

# 5. Điền biến vào prompt
promt = user_prompt.invoke({'topic': user_input, 'role': role_input})
print("\n--- PROMPT GỬI ĐẾN AI ---")
print(promt)
print("\n" + "="*40 + "\n")

# 6. Gọi LLM qua structured_llm để nhận kết quả khớp với Pydantic class
result = structured_llm.invoke(promt)

# 7. In kết quả các trường đã định nghĩa trong class
print("--- KẾT QUẢ ĐÃ ĐƯỢC CẤU TRÚC HÓA (PYDANTIC OBJECT) ---")
print(f"Chủ đề: {result.topic_name}")
print(f"Tóm tắt: {result.summary}")
print(f"Các ý chính: {result.key_points}")
print(f"Mức độ: {result.expertise_level}")
```
