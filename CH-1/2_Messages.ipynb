# Message

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_ollama import ChatOllama

llm_ollama = ChatOllama(model = 'llama3.2:3b', temperature=0)

my_message = [
    SystemMessage(content='bạn là 1 techlead 10 năm kinh nghiệm'),
    HumanMessage(content='Hãy nói cho tôi về machine learning ngắn gọn dễ hiểu')
]

llm_ollama.invoke(my_message).content
```

# Prompt

```python
from langchain_core.prompts import ChatPromptTemplate

user_prompt = ChatPromptTemplate([
    ('system','bạn là {role}'),
    ('user', 'hãy nói cho tôi về {topic}')
])
role_input = input("nhập vai trò: ")
user_input = input('nhập chủ đề muốn hỏi : ')
promt = user_prompt.invoke({'topic':user_input, 'role': role_input})
print(promt)
llm_ollama.invoke(promt).content
```
