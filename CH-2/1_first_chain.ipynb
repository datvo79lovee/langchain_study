```python
from langchain_ollama import ChatOllama
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

# CHAIN IN LANGCHAIN


# FIRST CHAIN

```python
# TASK 1- prompt 
prompt_template = ChatPromptTemplate.from_messages([
    ('system', 'Bạn là 1 chuyên gia'),
    ('human', '{message}')
])
```

```python
# TASK 2 - LLM
llm = ChatOllamallm_ollama = ChatOllama(model = 'llama3.2:3b', temperature=0)
```

```python
# TASK 3 - OUTPUT PARSER
from langchain_core.output_parsers import StrOutputParser
str_parser = StrOutputParser()
```

## Manual Ivocation

```python
template = prompt_template.invoke({
'message':'thủ đô của Việt Nam là gì?',
})

result = llm.invoke(template)
final_result = result.content
print(final_result)
```

## Chain Ivocation

```python
chain = prompt_template | llm | str_parser
chain.invoke({'message':'thủ đô của Việt Nam là gì?'})
```
